# ASCII Stream Engine (Demo)

Este notebook levanta el motor modular y transmite ASCII por UDP.

**Pasos:**
1. Instala dependencias: `python -m pip install -r ascii_stream_engine/requirements.txt`
2. Verifica que tengas `ffmpeg` instalado.
3. Abre VLC y carga: `udp://@127.0.0.1:1234`
4. Ejecuta las celdas en orden.

> Nota: cambios en `grid_w`, `grid_h`, `host` o `port` requieren reiniciar el motor.


```python
from ascii_stream_engine import (
    EngineConfig,
    StreamEngine,
    OpenCVCameraSource,
    AsciiRenderer,
    FfmpegUdpOutput,
    FilterPipeline,
    AnalyzerPipeline,
    build_control_panel,
)
from ascii_stream_engine.filters import BrightnessFilter, InvertFilter
from ascii_stream_engine.analyzers import FaceHaarAnalyzer

config = EngineConfig(host="127.0.0.1", port=1234, fps=20, grid_w=120, grid_h=60)
filters = FilterPipeline([BrightnessFilter(), InvertFilter()])
analyzers = AnalyzerPipeline([FaceHaarAnalyzer()])

engine = StreamEngine(
    source=OpenCVCameraSource(0),
    renderer=AsciiRenderer(),
    sink=FfmpegUdpOutput(),
    config=config,
    analyzers=analyzers,
    filters=filters,
)
```

```python
build_control_panel(engine)
```

```python
engine.start()
```

```python
engine.stop()
```
